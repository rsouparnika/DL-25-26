```python
import numpy as np

def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)

# XOR input
X = np.array([[0,0],[0,1],[1,0],[1,1]])

# Expected output
y = np.array([[0],[1],[1],[0]])

np.random.seed(0)

# Weight initialization
W1 = np.random.uniform(-1,1,(2,4))
W2 = np.random.uniform(-1,1,(4,1))

lr = 0.1

for epoch in range(20000):

    # Forward propagation
    hidden_layer = sigmoid(X @ W1)
    output_layer = sigmoid(hidden_layer @ W2)

    # Error
    error = y - output_layer

    # Backpropagation
    d_output = error * sigmoid_derivative(output_layer)
    d_hidden = (d_output @ W2.T) * sigmoid_derivative(hidden_layer)

    # Weight updates
    W2 += hidden_layer.T @ d_output * lr
    W1 += X.T @ d_hidden * lr

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error)):.4f}")

print("\nFinal Output after training:")
print(output_layer.round(3))
```

```python
import numpy as np

# Activation functions
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - x**2

# XOR input
X = np.array([[0,0],[0,1],[1,0],[1,1]])

# Expected output
y = np.array([[0],[1],[1],[0]])

np.random.seed(0)

# Weight initialization
W1 = np.random.uniform(-1,1,(2,4))
W2 = np.random.uniform(-1,1,(4,1))

lr = 0.1

for epoch in range(20000):

    # Forward propagation
    hidden_input = X @ W1
    hidden_layer = relu(hidden_input)

    output_input = hidden_layer @ W2
    output_layer = tanh(output_input)

    # Error
    error = y - output_layer

    # Backpropagation
    d_output = error * tanh_derivative(output_layer)
    d_hidden = (d_output @ W2.T) * relu_derivative(hidden_layer)

    # Weight updates
    W2 += hidden_layer.T @ d_output * lr
    W1 += X.T @ d_hidden * lr

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error)):.4f}")

print("\nFinal Output after training:")
print(output_layer.round(3))
```
